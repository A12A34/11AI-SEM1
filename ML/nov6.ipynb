```python
# Import utility to generate synthetic linear regression datasets
from sklearn.datasets import make_regression
import os
from pathlib import Path
BASE_DIR = Path(os.getcwd())
```

```python
# Generate a perfectly linear regression dataset (no noise)
# Parameters:
#   n_samples=100      -> number of rows (observations)
#   n_features=3       -> total feature columns
#   n_informative=3    -> all features contribute to target
#   n_targets=1        -> single regression output
#   noise=0.0          -> no random noise added (ideal linear relation)
#   coef=True          -> also return the true underlying coefficients
#   random_state=1     -> reproducible results
features, target, coefficients = make_regression(
    n_samples=100,
    n_features=3,
    n_informative=3,
    n_targets=1,
    noise=0.0,
    coef=True,
    random_state=1
)
```

```python
# Peek at the first few rows to verify shapes and values
print("features matrix \n", features[:3])
print("target vector \n", target[:3])
```

```python
# Import utility to generate synthetic classification datasets
from sklearn.datasets import make_classification
```

```python
# Generate an imbalanced binary classification dataset
# Parameters:
#   n_samples=100     -> number of rows (observations)
#   n_features=3      -> total features per sample
#   n_informative=3   -> all features carry signal for the class label
#   n_redundant=0     -> no linear combinations of informative features
#   n_classes=2       -> binary target (0/1)
#   weights=[.25,.75] -> class distribution: 25% of class 0, 75% of class 1
#   random_state=1    -> reproducible results
features, target = make_classification(n_samples=100,
    n_features=3,
    n_informative=3,
    n_redundant=0,
    n_classes=2,
    weights=[.25, .75],
    random_state=1,
)
```

```python
# Display a quick preview of the classification feature rows and target labels
print("Features (first 4 rows):\n", features[:4])
print("Target (first 3 labels):\n", target[:3])
```
