```python
# Import utility to generate synthetic clustering datasets (Gaussian blobs)
from sklearn.datasets import make_blobs
```

```python
# Create a simple 2D clustering dataset (3 Gaussian blobs)
# Parameters:
#   n_samples=100   -> total points to generate
#   n_features=2    -> 2D features, easy to visualize
#   centers=3       -> number of clusters
#   cluster_std=0.5 -> cluster spread (smaller -> tighter clusters)
#   shuffle=True    -> shuffle the samples
#   random_state=1  -> reproducible results
features, target = make_blobs(
    n_samples=100,
    n_features=2,
    centers=3,
    cluster_std=0.5,
    shuffle=True,
    random_state=1,
)
```

```python
print("feature:\n",features[:5])
print("target:\n",target[:5])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(
    features[:, 0],          # x-axis values (first feature)
    features[:, 1],          # y-axis values (second feature)
    c=target,                # color by cluster label
)
plt.show()
```
